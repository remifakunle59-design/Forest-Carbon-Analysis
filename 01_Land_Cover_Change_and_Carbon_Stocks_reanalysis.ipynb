# Land-cover change and carbon-stock reanalysis (1985–2020)

**Research questions:** How did mapped land-cover areas change? How do estimated carbon stocks differ across land uses? What stock differences follow when class-average carbon densities are applied to each map year?

**Sources:** MSc thesis Table 4.2 (land-cover areas), Table 4.7 (field inventory), and soil-related Tables 4.6, 4.8 and 4.11. Values are transcribed below; confirm them against the thesis before publication.

**Scope:** These are *area-based stock estimates*, not independently measured historical carbon stocks, mapped land-use transitions, or observed ecosystem CO₂ fluxes. Class-average carbon densities are held constant over time.


## Data, assumptions and important distinctions

- Forty field plots cover six field land uses. Aboveground and belowground values are paired at the plot level.
- The `soil` column in the embedded plot table is **not** the source used for final 0–30 cm soil-carbon density. Instead, the separate `soil_carbon` series is used, reportedly from depth-aggregated thesis soil calculations. **Verify this distinction and the units against the original tables.** Do not interpret the plot `soil` column as interchangeable.
- Soil uncertainty is an *approximation* propagated from soil organic-carbon concentration and bulk-density variation. It omits depth/thickness error and possible covariance.
- Aboveground and belowground carbon covary. Their *paired plot sums* are used for their combined variability, rather than incorrectly treating these two pools as independent.
- Soil uncertainty is combined with vegetation uncertainty **assuming independence**, because paired soil-density estimates and covariance are not available in this notebook. Error bars are illustrative, not full confidence intervals.
- Class carbon density is held fixed for all historical years. Map classification errors, within-class temporal changes and boundary inconsistencies are not propagated.
- Urban, rock and water are set to zero for this simplified accounting exercise; this does **not** mean those classes contain no carbon.
- Oil palm and cocoa are grouped as plantation; cassava and maize as cropland.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from io import StringIO
#Carbon_Stock_Across_6_land_use_types_within_3_carbon_pool
csv_text = """land_use_types,aboveground,belowground,soil
Dense Forest,116.88,27.47,9.91
Dense Forest,117.23,27.55,11.45
Dense Forest,120.95,28.42,15.5
Dense Forest,112.71,26.49,14.74
Dense Forest,118.72,27.9,11.8
Dense Forest,129.75,30.49,29.15
Light Forest,100.2,23.55,10.055
Light Forest,82.91,19.48,12.247
Light Forest,86.45,20.32,14.132
Light Forest,82.75,19.45,11.59
Light Forest,115.82,27.22,17.025
Light Forest,75.42,17.72,13.774
Oil Palm,72.8,17.11,8.64
Oil Palm,78.83,18.53,9.84
Oil Palm,71.95,16.91,7.17
Oil Palm,61.12,14.36,9.63
Oil Palm,53.59,12.59,7.55
Oil Palm,65.15,15.31,6.69
Oil Palm,55.43,13.03,7.42
Oil Palm,79.59,18.7,6.39
Cocoa,49.25,11.57,7.41
Cocoa,53.73,12.63,7.46
Cocoa,60.13,14.13,6.76
Cocoa,61.12,14.36,9.89
Cocoa,58.3,13.7,7.1
Cocoa,85.88,20.18,7.18
Cocoa,72.9,17.13,9
Cocoa,59.34,13.94,6.23
Cassava,17.77,4.18,7.89
Cassava,19.64,4.61,8.63
Cassava,20.99,4.93,9.07
Cassava,18.72,4.4,9.71
Cassava,15.93,3.74,6.92
Cassava,23.21,5.46,9.66
Maize,10.03,2.36,6.31
Maize,13.98,3.29,7.4
Maize,11.57,2.72,9.24
Maize,16.83,3.96,8.34
Maize,11.22,2.64,8.66
Maize,15.21,3.57,10.19
"""

plots = pd.read_csv(StringIO(csv_text))

order = ["Dense Forest", "Light Forest", "Oil Palm", "Cocoa", "Cassava", "Maize"]
n = plots["land_use_types"].value_counts().reindex(order)

# Aboveground and belowground carbon: mean and std from the plot values
ag = plots.groupby("land_use_types")["aboveground"].agg(["mean", "std"]).reindex(order)
bg = plots.groupby("land_use_types")["belowground"].agg(["mean", "std"]).reindex(order)

# Soil carbon, 0-30 cm (Mg C/ha)
soil_carbon = pd.Series([92.55, 83.82, 63.33, 61.02, 51.88, 50.15], index=order)

# Soil uncertainty:
soil_conc_mean = pd.Series([1.50, 1.40, 1.50, 1.35, 1.59, 1.77], index=order)
soil_conc_std   = pd.Series([0.48, 0.69, 0.53, 0.54, 0.76, 1.09], index=order)
BD_mean   = pd.Series([1.35, 1.44, 1.48, 1.51, 1.55, 1.58], index=order)
BD_std     = pd.Series([0.09, 0.10, 0.10, 0.08, 0.09, 0.08], index=order)
soil_std = soil_carbon * np.sqrt((soil_conc_std / soil_conc_mean) ** 2 + (BD_std / BD_mean) ** 2)

carbon = pd.DataFrame({
    "n": n,
    "aboveground": ag["mean"], "belowground": bg["mean"], "soil": soil_carbon,
})
carbon["total"] = carbon[["aboveground", "belowground", "soil"]].sum(axis=1)

# Paired aboveground + belowground values retain their observed covariance.
veg_total = (plots.assign(vegetation=plots["aboveground"] + plots["belowground"])
             .groupby("land_use_types")["vegetation"]
             .agg(["mean", "std"]).reindex(order))
# Soil uncertainty is approximate and assumed independent of vegetation.
carbon["total_std"] = np.sqrt(veg_total["std"] ** 2 + soil_std ** 2)
print("CHECK: soil_carbon comes from separately aggregated soil calculations;")
print("the embedded plot-level soil column is NOT used in total carbon.")

print("CARBON STOCK (Mg C/ha)")
print(carbon.round(1))

plots.to_csv("table_4_7_plots.csv", index=False)
```

## Indicative uncertainty (not a complete confidence interval)

Aboveground and belowground biomass are paired measurements, so their observed covariance is retained by calculating the standard deviation of the **plot-level vegetation sum**. The separate soil uncertainty estimate is combined in quadrature with vegetation variability, **assuming soil and vegetation are independent**. Dividing by √n gives an illustrative standard-error scale. Soil inputs and the assumptions must be verified against the thesis before treating these error bars as inferential uncertainty.


```python
import numpy as np
import matplotlib.pyplot as plt

SE_total = carbon["total_std"] / np.sqrt(carbon["n"])

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(carbon.index, carbon["total"], yerr=SE_total, capsize=6, color="red")
ax.set_ylabel("Total carbon stock (Mg C/ha)")
ax.set_title("Total carbon stock by land use (mean +/- SE)")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

## Changes in land-cover **area**, 1985–2020

These are marginal class-area totals. They quantify net changes in each class but **cannot identify which individual pixels changed from one class to another**. A land-cover transition matrix would require aligned classified rasters.


```python
import pandas as pd
years = [1985, 1992, 1999, 2006, 2013, 2020]
area_km2 = {
    "Dense Forest": [2491.04, 2035.48, 1586.54, 1240.47, 833.29, 592.61],
    "Light Forest": [296.30, 611.99, 768.98, 806.25, 733.52, 650.32],
    "Plantation":   [354.46, 471.43, 734.98, 979.53, 1431.53, 1586.17],
    "Cropland":     [40.19, 58.06, 72.41, 96.81, 98.13, 67.98],
    "Urban Area":   [58.00, 63.05, 77.09, 116.93, 143.53, 342.92],
    "Rock Outcrop": [18.69, 18.44, 18.01, 17.86, 16.35, 14.01],
    "Water Bodies": [25.05, 24.09, 23.78, 21.42, 19.86, 17.54],
}
lulc = pd.DataFrame(area_km2, index=years).T

lulc["change_km2"] = lulc[2020] - lulc[1985]
lulc["change_pct"] = lulc["change_km2"] / lulc[1985] * 100
lulc["share_1985_pct"] = lulc[1985] / lulc[1985].sum() * 100
lulc["share_2020_pct"] = lulc[2020] / lulc[2020].sum() * 100

print(lulc[[1985, 2020, "change_km2", "change_pct",
            "share_1985_pct", "share_2020_pct"]].round(1))

lulc[years].T.plot(figsize=(8, 5), marker="o")
plt.ylabel("Area (km2)")
plt.title("Land-cover change, 1985-2020")
plt.tight_layout()
plt.show()
print("\nMapped total area by year (km²):")
print(lulc[years].sum(axis=0).round(2))
print("Check against thesis study-area definition and the boundary used in the satellite notebook.")
```

## Upscaling to landscape carbon-stock estimates

Each mapped class area (km² × 100 = hectares) is multiplied by a fixed mean carbon density (Mg C/ha). The result is expressed in Tg C. Grouped plantation and cropland estimates use plot-count-weighted summaries; this is a **sampling assumption**, not an area-weighted representation of oil palm versus cocoa or cassava versus maize. Reported error estimates exclude map uncertainty, sampling design effects and possible covariance among classes.


```python
def pool(rows):
    # Combine classes: rows = [(n, mean, sd), ...] -> (n, mean, sd)
    N = sum(r[0] for r in rows)
    M = sum(r[0] * r[1] for r in rows) / N
    var = (sum((r[0] - 1) * r[2] ** 2 for r in rows)
           + sum(r[0] * (r[1] - M) ** 2 for r in rows)) / (N - 1)
    return N, M, np.sqrt(var)

def row(name):
    return (carbon.loc[name, "n"], carbon.loc[name, "total"], carbon.loc[name, "total_std"])

groups = {
    "Dense Forest": [row("Dense Forest")],
    "Light Forest": [row("Light Forest")],
    "Plantation":   [row("Oil Palm"), row("Cocoa")],
    "Cropland":     [row("Cassava"), row("Maize")],
}
stats = {}
for name, rows in groups.items():
    N, M, S = pool(rows)
    stats[name] = (M, S / np.sqrt(N))          # mean (Mg C/ha) and its standard error
stats = pd.DataFrame(stats, index=["mean", "se"]).T
print(stats.round(2))

classes = ["Dense Forest", "Light Forest", "Plantation", "Cropland"]
area_ha = lulc.loc[classes, years].T * 100          # km2 -> ha

stock = area_ha.mul(stats["mean"], axis=1) / 1e6    # Tg C (million tonnes)
err = area_ha.mul(stats["se"], axis=1) / 1e6

total = stock.sum(axis=1)
total_err = np.sqrt((err ** 2).sum(axis=1))

result = pd.DataFrame({"carbon_Tg": total, "uncertainty_Tg": total_err})
print(result.round(1))
print("Change 1985-2020:", round(total[2020] - total[1985], 1), "Tg C,",
      round((total[2020] / total[1985] - 1) * 100, 1), "%")

fig, ax = plt.subplots(figsize=(8, 5))
stock.plot(kind="bar", stacked=True, ax=ax,
           color=["green", "purple", "brown", "orange"])
ax.set_ylabel("Carbon stock (Tg C)")
ax.set_title("Carbon stock by land-cover class, 1985-2020")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Annualised stock difference (not measured flux)

Differences between estimated stocks at successive map dates are divided by the number of intervening years. Multiplication by 44/12 converts a **carbon-stock difference** into a CO₂ mass equivalent. It does not establish the timing of emissions, atmospheric exchange, or a measured net CO₂ flux. Negative values represent declines in estimated stocks.


```python
yrs = pd.Series(years, index=years)
rate = total.diff() / yrs.diff()     # Tg C per year
co2e = rate * 44 / 12                # Tg CO2e per year

print(pd.DataFrame({"C_change_Tg_per_yr": rate,
                    "CO2e_Tg_per_yr": co2e}).round(2))
```

## Interpretation and limitations

- Dense forest *mapped area* declined by approximately 76% between 1985 and 2020.
- Plantation and light-forest mapped areas increased; their assumed carbon densities contribute substantially to estimated landscape carbon.
- Total **estimated** landscape carbon stock declined under constant class-density assumptions.
- The analysis cannot attribute specific forest-to-plantation or forest-to-cropland transitions without classified spatial maps.
- Major uncertainties: limited and potentially non-representative plots; separately aggregated soil data; approximated soil uncertainty; fixed historical class carbon densities; missing classification and area uncertainty; simplified zero-carbon classes; and potentially different study-area boundaries.
- These results describe estimated **stocks and stock differences**, not measured carbon uptake, a forest sink, or direct atmospheric CO₂ exchange.
